```python

from pathlib import Path
import pandas as pd
import numpy as np

# Go from notebooks/ to the project root
PROJECT_ROOT = Path.cwd().parent

# Build the dataset path
DATA_PATH = PROJECT_ROOT / "data" / "raw" / "dataset.csv"

print("Project root:", PROJECT_ROOT)
print("Dataset path:", DATA_PATH)
print("File exists:", DATA_PATH.exists())

# Load the dataset
if not DATA_PATH.exists():
    raise FileNotFoundError(
        f"Dataset not found at {DATA_PATH}\n"
        "Verify that dataset.csv exists inside data/raw/."
    )

df = pd.read_csv(DATA_PATH)

print("\nDataset loaded successfully!")
print("Shape:", df.shape)

```

```python
# Total missing values across the dataset
print(df.isnull().sum())

# Total number of missing cells
print("Total missing values:", df.isnull().sum().sum())

# Missing values by column, sorted highest first
print(df.isnull().sum().sort_values(ascending=False))

# Percentage of missing values per column
missing_percentage = (df.isnull().mean() * 100).sort_values(
    ascending=False
)
print(missing_percentage)
```

```python
missing = df.isnull().sum()
print(missing[missing > 0])
```

```python
# Summary of numerical columns
df.describe()

# Include all columns, including categorical ones
df.describe(include="all")
```

```python
# Number of unique values in each column
print(df.nunique().sort_values())

# Unique values in a specific column
print(df["SepsisLabel"].unique())

# Check numerical ranges
print(df.select_dtypes(include="number").min())
print(df.select_dtypes(include="number").max())
```
